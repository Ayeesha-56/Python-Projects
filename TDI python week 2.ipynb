```python
## Task 2: Process Multiple Orders (FOR LOOPS)
order_values = [12000, 55000, 150000, 49999, 100001, 8000]

def approval_message(order_value):
    if order_value > 100000:
        return "Needs manager approval"
    elif order_value > 50000:
        return "Needs supervisor approval"
    else:
        return "Auto-approved"

for order_value in order_values:
    print(order_value, approval_message(order_value))
```

```python
## Task 3: Supplier Status Function
suppliers = [
    ("SupplierA", 0.94),
    ("SupplierB", 0.86),
    ("SupplierE", 0.74)
]

def check_supplier_status(supplier_name, on_time_rate):
    if on_time_rate >= 0.90:
        return 'Reliable'
    elif on_time_rate >= 0.80:
        return 'Monitor'
    else:
        return 'At Risk'

for supplier_name, on_time_rate in suppliers:
    print(supplier_name, check_supplier_status(supplier_name, on_time_rate))
```

```python
##Task 4: Load the Supply Chain Dataset
import csv
with open("supply_chain_dataset.csv", newline="") as file:
 reader = csv.DictReader(file)
 rows = list(reader)
```

```python
# Print first 5 rows
print("First 5 rows:")
print(rows[:5])
```

```python
## How many different supplier names are written in the file?
suppliers = set()

for row in rows:
    suppliers.add(row["supplier_name"])

print("Number of different suppliers:", len(suppliers))
print("Supplier names:", suppliers)
```

```python
## What is the highest total_cost?
highest_cost = 0

for row in rows:
    cost = float(row["total_cost"])

    if cost > highest_cost:
        highest_cost = cost

print("Highest total cost:", highest_cost)
```

```python
## What is the lowest total_cost?
lowest_cost = float(rows[0]["total_cost"])

for row in rows:
    cost = float(row["total_cost"])

    if cost < lowest_cost:
        lowest_cost = cost

print("Lowest total cost:", lowest_cost)
```

```python
## How many orders have the delivery_status "Delivered"?
delivered_count = 0

for row in rows:
    if row["delivery_status"] == "Delivered":
        delivered_count += 1

print("Number of delivered orders:", delivered_count)
```

```python
## Challenge 1: Write a function that takes an order value and a supplier's on-time
##rate, and returns a different approval level based on both. For example, reject
##orders above 100,000 from suppliers below 0.80.

def approval_level(order_value, on_time_rate):

    if order_value > 100000 and on_time_rate < 0.80:
        return "Rejected"

    elif order_value > 100000:
        return "Needs manager approval"

    elif order_value > 50000:
        return "Needs supervisor approval"

    else:
        return "Auto-approved"

print(approval_level(150000, 0.74))
print(approval_level(150000, 0.94))
print(approval_level(72500, 0.86))
print(approval_level(12000, 0.94))
```

```python
## Find all orders from SupplierA, however the name is written. Hint:
# name.lower().replace(" ", "") turns "Supplier A" into "suppliera". Count them
# and add up their total cost.

supplier_count = 0
total_cost = 0


# Step 3: Loop through all rows
for row in rows:

    # Get supplier name
    name = row["supplier_name"]

    # Standardise supplier name and check if it is SupplierA
    if name.lower().replace(" ", "") == "suppliera":

        # Count SupplierA orders
        supplier_count += 1

        # Add up SupplierA order costs
        total_cost += float(row["total_cost"])


# Step 4: Print results
print("Number of SupplierA orders:", supplier_count)
print("Total cost of SupplierA orders:", total_cost)
```

```python
## Loop through the dataset and find rows with empty values
#(row[column] == ""). Which columns have gaps, and how many?
# Create an empty dictionary
missing_values = {}

# Initialise each column count to zero
for column in rows[0].keys():
    missing_values[column] = 0


# Check every row and every column
for row in rows:
    for column in row.keys():
        if row[column] == "":
            missing_values[column] += 1


# Print columns with missing values
print("Columns with missing values:")

for column, count in missing_values.items():
    if count > 0:
        print(column, ":", count)
```
